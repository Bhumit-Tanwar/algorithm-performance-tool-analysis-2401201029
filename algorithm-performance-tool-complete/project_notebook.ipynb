```python
import pandas as pd
df = pd.read_csv('benchmark_results.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
for algo in df['algorithm'].unique():
    s = df[df.algorithm == algo]
    plt.plot(s.input_size, s.time_sec, marker='o', label=algo)
plt.xlabel('Input Size (n)'); plt.ylabel('Execution Time (seconds)')
plt.title('Execution Time Comparison'); plt.legend(); plt.grid(True); plt.show()
```

## Observation
Binary search grows much more slowly than linear search. Nested loops show the strongest growth as input size increases. Recursive factorial uses additional call-stack space, while iterative factorial uses constant auxiliary space.
